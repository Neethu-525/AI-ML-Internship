# Section C: Mini Task

Create program:

.Take 5 student inputs  
• Store in list  
• Find average  
• Find topper  

```python
students = []

# 1. Take 5 student inputs
for i in range(5):
    print(f"\n--- Student {i + 1} ---")
    name = input("Enter student name: ")
    marks = float(input("Enter marks: "))

    # Store as a dictionary inside the list
    students.append({"name": name, "marks": marks})

# 2. Find the average marks using a loop
total_marks = 0
for student in students:
    total_marks += student["marks"]

average_marks = total_marks / len(students)

# 3. Find the topper (student with the highest marks)
topper = students[0] # Assume the first student is the topper
for student in students:
    if student["marks"] > topper["marks"]:
        topper = student

# 4. Print the results
print("\n=== Results ===")
print(f"Class Average: {average_marks:.2f}")
print(f"Topper: {topper['name']} with {topper['marks']} marks")
```

# function

1. function to find factorial

```python
def factorial(n):
  fact = 1

  for i in range(1, n+1):
    fact = fact*i

  return fact

num = 5
answer = factorial(num)
print(answer)
```

2. function to check prime number

```python
def check_prime(num):
  if num > 1:
    for i in range(2,num):
      if num % i == 0:
        return "not prime"
      return "prime"
  else:
    return "not prime"
number = int(input("enter the number :"))
result = check_prime(number)
print(result)
```

# Task 6: Logic Challenge

Find sum of digits using recursion  

```python
def sum_of_digits(n):
    if n < 10:
        return n
    return (n % 10) + sum_of_digits(n // 10)
num = 12345
result = sum_of_digits(num)

print(f"The sum of the digits of {num} is: {result}")
```

Flatten nested list  

```python
def flatten_list(nested_list):
    flat_list = []

    for item in nested_list:
        if isinstance(item, list):
            flat_list.extend(flatten_list(item))
        else:
            flat_list.append(item)

    return flat_list
nested = [1, [2, 3], [4, [5, 6]], 7]
result = flatten_list(nested)

print("Original list:", nested)
print("Flattened list:", result)
```

function to find topper

```python
def find_topper(student_list):
    if not student_list:
        return None

    topper = max(student_list, key=lambda student: student["marks"])
    return topper
students = [
    {"name": "Alice", "marks": 85},
    {"name": "Bob", "marks": 72},
    {"name": "Charlie", "marks": 90},
    {"name": "Diana", "marks": 60}
]

topper_student = find_topper(students)
print(f"The topper is {topper_student['name']} with {topper_student['marks']} marks!")
```

function to filter students

```python
def filter_students(student_list, condition_func):
    return [student for student in student_list if condition_func(student)]

students = [
    {"name": "Alice", "marks": 85, "grade": "A"},
    {"name": "Bob", "marks": 72, "grade": "B"},
    {"name": "Charlie", "marks": 90, "grade": "A"},
    {"name": "Diana", "marks": 60, "grade": "C"}
]
top_students = filter_students(students, lambda s: s["marks"] > 75)
print("Marks > 75:", top_students)
grade_a_students = filter_students(students, lambda s: s["grade"] == "A")
print("Grade A:", grade_a_students)
d_students = filter_students(students, lambda s: s["name"].startswith("D"))
print("Names starting with D:", d_students)
```

1. create method to:
     check pass/fail
     calculate grade

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Method to check pass or fail
    def check_result(self):
        if self.marks >= 40:
            return "Pass"
        else:
            return "Fail"

    # Method to calculate grade
    def calculate_grade(self):
        if self.marks >= 90:
            return "A+"
        elif self.marks >= 75:
            return "A"
        elif self.marks >= 60:
            return "B"
        elif self.marks >= 40:
            return "C"
        else:
            return "F"
student1 = Student("Sivani", 82)

print("Name:", student1.name)
print("Marks:", student1.marks)
print("Result:", student1.check_result())
print("Grade:", student1.calculate_grade())
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def check_result(self):
        if self.marks >= 40:
            return "Pass"
        else:
            return "Fail"

    def calculate_grade(self):
        if self.marks >= 90:
            return "A+"
        elif self.marks >= 75:
            return "A"
        elif self.marks >= 60:
            return "B"
        elif self.marks >= 40:
            return "C"
        else:
            return "F"


# Create list of student objects
students = [
    Student("Anu", 85),
    Student("Rahul", 92),
    Student("Meera", 67),
    Student("Arjun", 45),
    Student("Neha", 30)
]

# Print all students
print("Student Details:\n")

for s in students:
    print("Name:", s.name)
    print("Marks:", s.marks)
    print("Result:", s.check_result())
    print("Grade:", s.calculate_grade())
    print("---")

# Find topper
topper = students[0]

for s in students:
    if s.marks > topper.marks:
        topper = s

print("Topper:")
print("Name:", topper.name)
print("Marks:", topper.marks)
```
